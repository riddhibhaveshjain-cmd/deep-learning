```python
image_root = r'C:\AI\deep learning\agedetectiontrain'

image_paths = {}
for root, _, files in os.walk(image_root):
    for filename in files:
        image_paths.setdefault(filename, os.path.join(root, filename))

temp = []
for img_name in train.ID:
    filename = os.path.basename(str(img_name).strip())
    img_path = image_paths.get(filename)

    if img_path is None:
        valid_rows = []
        temp = []
        missing_files = []

        for row_idx, img_name in train.ID.items():
            filename = os.path.basename(str(img_name).strip())
            img_path = image_paths.get(filename)

            if img_path is None:
                missing_files.append(filename)
                continue

            try:
                img = imageio.imread(img_path)
                img = np.array(
                    Image.fromarray(img).resize((32, 32))
                ).astype('float32')
                temp.append(img)
                valid_rows.append(row_idx)
            except Exception as exc:
                missing_files.append(f'{filename}: {exc}')

        train = train.loc[valid_rows].reset_index(drop=True)
        train_x = np.stack(temp)

        print(f'Skipped {len(missing_files)} missing or unreadable files.')
        print('Examples:', missing_files[:10])

    img = imageio.imread(img_path)
    img = np.array(Image.fromarray(img).resize((32, 32))).astype('float32')
    temp.append(img)

train_x = np.stack(temp)
```
